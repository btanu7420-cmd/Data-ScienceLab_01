```python
import pandas as pd
import numpy as np
```

```python
# 1. Create a TSV file with at least 8 students

data = {
    "StudentID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Aman", "Priya", "Rahul", "Sneha", "Vikas", "Neha", "Arjun", "Kriti"],
    "Department": ["CSE", "ECE", "CSE", "ME", "ECE", "CSE", "ME", "ECE"],
    "Attendance": [90, 85, 78, 88, 92, 81, 87, 95],
    "Marks": [88, 76, 91, 69, 84, 73, 80, 94]
}

df = pd.DataFrame(data)

# Save as TSV file
df.to_csv("students.tsv", sep="\t", index=False)

print("TSV file created successfully!\n")
```

```python
# 2. Read the TSV file and display first 3 rows

df = pd.read_csv("students.tsv", sep="\t")

print("First 3 Rows:")
print(df.head(3))
```

```python
# 3. Find number of rows and columns

rows, cols = df.shape
print("\nNumber of Rows:", rows)
print("Number of Columns:", cols)
```

```python
# 4. Display only Name and Marks

print("\nName and Marks:")
print(df[["Name", "Marks"]])
```

```python
# 5. Filter students with Attendance >= 85

print("\nStudents with Attendance >= 85:")
print(df[df["Attendance"] >= 85])
```

```python
# 6. Count students in each department

print("\nStudent Count by Department:")
print(df["Department"].value_counts())
```

```python
# 7. Average marks for each department

print("\nAverage Marks by Department:")
print(df.groupby("Department")["Marks"].mean())
```

```python
# 8. Add missing value and handle it in two ways
# Add a missing value
df.loc[2, "Marks"] = np.nan

print("\nData with Missing Value:")
print(df)

# Method 1: Fill missing values with mean
df_fill = df.copy()
df_fill["Marks"] = df_fill["Marks"].fillna(df_fill["Marks"].mean())

print("\nMethod 1: Filled Missing Value with Mean")
print(df_fill)

# Method 2: Drop rows with missing values
df_drop = df.dropna()

print("\nMethod 2: Dropped Missing Value Rows")
print(df_drop)
```

```python
# 9. Save cleaned data (filled version) to new TSV

df_fill.to_csv("students_cleaned.tsv", sep="\t", index=False)

print("\nCleaned data saved as 'students_cleaned.tsv'")
```

```python
# 10. Read cleaned file and verify columns and shape

cleaned_df = pd.read_csv("students_cleaned.tsv", sep="\t")

print("\nColumn Names:")
print(cleaned_df.columns.tolist())

print("\nShape of Cleaned Data:")
print(cleaned_df.shape)
```
